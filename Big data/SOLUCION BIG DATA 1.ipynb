```python
import time
from pyspark.sql import functions as F
```

```python
ruta = "/Volumes/workspace/distribuidora_los_andes/datos_proyecto/entrada/productos.csv"
```

```python
t_inicio = time.time()
df = spark.read.csv(ruta, header=True, inferSchema=True)
total_filas = df.count()
t_lectura = round(time.time() - t_inicio, 3)

print(f"Total registros cargados: {total_filas} | Tiempo de lectura: {t_lectura} s")
```

```python
t0 = time.time()
duplicados = df.groupBy("id_producto").count().filter(F.col("count") > 1).count()
t_dup = round(time.time() - t0, 3)

print(f"Duplicados en id_producto: {duplicados} | Tiempo: {t_dup} s")
```

```python
t0 = time.time()
nulos = df.filter(
    F.col("id_producto").isNull() |
    F.col("categoria").isNull() |
    F.col("precio_unitario_soles").isNull() |
    F.col("stock_disponible").isNull()
).count()
t_nul = round(time.time() - t0, 3)

print(f"Filas con nulos en campos clave: {nulos} | Tiempo: {t_nul} s")
```

```python
t0 = time.time()
invalidos = df.filter(
    (F.col("precio_unitario_soles") <= 0) |
    (F.col("stock_disponible") < 0)
).count()
t_inv = round(time.time() - t0, 3)

tiempo_total = round(t_lectura + t_dup + t_nul + t_inv, 3)
rendimiento = round(total_filas / tiempo_total, 1) if tiempo_total > 0 else 0

print(f"Registros inválidos (precio <= 0 o stock < 0): {invalidos} | Tiempo: {t_inv} s")
print(f"Tiempo acumulado: {tiempo_total} s | Rendimiento: {rendimiento} registros/s")
```
